```python
import sqlite3
import os
from ipumspy import readers

xml_file = "../data/raw/cps_00001.xml" 
dat_gz_file = "../data/raw/cps_00001.dat.gz"
db_file = "../data/ipums_database.db"

conn = sqlite3.connect(db_file)

ddi = readers.read_ipums_ddi(xml_file)

reader = readers.read_microdata_chunked(ddi, dat_gz_file, chunksize=100000)

print("Streaming started... this might take a few minutes.")

for i, df in enumerate(reader):
    df.to_sql("cps_extract", conn, if_exists="append", index=False)
    print(f"Finished chunk {i+1}")

conn.close()
print("All done! Your .db file is ready.")
```
