```python
import os
import sys
import findspark

os.environ["JAVA_HOME"] = os.path.join(os.environ["CONDA_PREFIX"], "Library")

findspark.init()

from pyspark.sql import SparkSession

os.environ['PYSPARK_PYTHON'] = sys.executable
os.environ['PYSPARK_DRIVER_PYTHON'] = sys.executable

spark = SparkSession.builder \
    .appName("CPS_Analysis") \
    .master("local[*]") \
    .config("spark.driver.memory", "8g") \
    .getOrCreate()

print("Spark is live!")
```

```python
import pandas as pd

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)

df = spark.read.option("mergeSchema", "False").parquet("../data/cps_data.parquet")
df.limit(30).toPandas()
```

```python
print(df.count())
```

```python
from pyspark.sql.functions import isnan, when, count, col

nan_summary = df.select([count(when(isnan(c) | col(c).isNull(), c)).alias(c) for c in df.columns]).toPandas()
nan_summary
```

```python
target_cols = ['HFLAG','YRIMMIG','CITIZEN','UHRSWORKT','DIFFANY',\
               'EITCRED','HEALTH','OCC_HEAD','OCC_MOM','OCC_MOM2',\
               'OCC_POP','OCC_POP2','OCC_SP','EDUC_HEAD','EDUC_MOM',\
               'EDUC_MOM2','EDUC_POP','EDUC_POP2','EDUC_SP','INCTOT_HEAD',\
               'INCTOT_MOM','INCTOT_MOM2','INCTOT_POP','INCTOT_POP2',\
               'INCTOT_SP']

for c in target_cols:
    print(f"--- Non-null values for {c} ---")
    sample_values = df.filter(col(c).isNotNull()).select(c).distinct().limit(5).toPandas()
    print(sample_values)
    print("\n")
```

```python
for col_name in df.columns:
    print(f"=========================================")
    print(f"COLUMN: {col_name}")
    print(f"=========================================")
    
    unique_count = df.select(col_name).distinct().count()
    print(f"Total unique values: {unique_count}")
    
    if unique_count <= 20:
        print("Values & Counts:")
        freq_df = df.groupBy(col_name).count().orderBy(col_name).toPandas()
        print(freq_df.to_string(index=False))
        
    else:
        print("Top 10 Most Frequent Values (Potential Flags/Top-Codes):")
        freq_df = df.groupBy(col_name).count().orderBy(col("count").desc()).limit(10).toPandas()
        print(freq_df.to_string(index=False))
        
    print("\n")
```
